```python
import random 

class Creature():
    def __init__(self, name, health, attack_power):
        self.name = name
        self.health = health
        self.attack_power = attack_power
    
    def take_damage(self, damage):
        self.health = self.health - damage
        print(f"{self.name} takes {damage} damage")

    def attack(self, opponent):
        opponent.take_damage(self.attack_power)

    def is_alive(self):
        return True if self.health > 0 else False
    
    def __str__(self):
        return f"{self.name}: {self.health} HP"
```

```python
import random
import time 

def battle(creature1, creature2):
    print("="*10)
    print("The battle begins!")

    while creature1.is_alive() and creature2.is_alive():
        choice = random.choice(["H", "T"])
        if choice == "H":
            creature1.attack(creature2)
        else:
            creature2.attack(creature1)
        print(creature1)
        print(creature2)
        time.sleep(2)
        print("-------")

    if creature1.is_alive(): 
        print(f"{creature1.name} wins!")
    else:
        print(f"{creature2.name} wins!")
    print("Battle ended")
```

```python
knight = Creature("Knight", 200, 20)
archer = Creature("Archer", 120, 35)

battle(knight, archer)
```

# Add an armor to my "Knight" class using inheritance and crit chance to "Archer" class

Armor reduces damage by a flat amount every time knight is attacked

```python
class Knight(Creature):
    def __init__(self, name, health, attack_power, armor):
        super().__init__(name, health, attack_power)
        self.armor = armor
    
    def take_damage(self, damage):
        damage = max(damage - self.armor, 0) # Reduce damage by armor, but not below 0
        self.health = self.health - damage
        print(f"{self.name} took {damage} damage after armor reduction!")
```

```python
class Archer(Creature):
    def __init__(self, name, health, attack_power, crit_chance):
        super().__init__(name, health, attack_power)
        self.crit_chance = crit_chance
    
    def attack(self, opponent):
        if random.random() <= self.crit_chance:
            damage = self.attack_power * 2
            print(f"{self.name} lands a critical hit!")
        else:
            damage = self.attack_power
        opponent.take_damage(damage)
```

```python
creature1 = Knight("Knight", 200, 20, armor=10)
creature2 = Archer("Archer", 120, 30, crit_chance=0.2)
```

```python
battle(creature1, creature2)
```
