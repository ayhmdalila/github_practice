```python
class Creature():
    def __init__(self, name, health, attack_power):
        self.name = name
        self.health = health
        self.attack_power = attack_power
    
    def attack(self, opponent):
        opponent.health = opponent.health - self.attack_power
        print(f"{self.name} attacks {opponent.name} for {self.attack_power} damage!")
    
    def is_alive(self):
        return True if self.health>0 else False
    
    def __str__(self):      
        return f"{self.name}: {self.health} HP"
```

```python
import random
import time 

def battle(creature1, creature2): 
    print("="*10)
    print("The battle begins!")
    turns = 0
    while creature1.is_alive() and creature2.is_alive():
        random_choice = random.choice(["heads", "tails"])
        if random_choice == "heads":
            creature1.attack(creature2)
        else:
            creature2.attack(creature1)
        print("-------")
        print(f"{creature1.name}: {creature1.health} HP")
        print(f"{creature2.name}: {creature2.health} HP")
        print("\n")
        time.sleep(2) # Sleep for 1 second between each turn
        turns = turns + 1
    print("="*10)
    print(f"Battle is over after {turns} turns!")
    print("Winner is: " + creature1.name if creature1.is_alive() else creature2.name)
    print("="*10)
```

```python
knight = Creature("Knight", 200, 20)
archer = Creature("Archer", 120, 35)

battle(knight, archer)
```
